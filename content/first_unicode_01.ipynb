```python
from unicodedata import name, category
```

```python
def uname(c: str) -> str:
    try:
        n = name(c)
    except ValueError:
        n = ""
    finally:
        return n
```

```python
def allu(i :int):
    c = chr(i)
    ca = category(c)
    if not c.isprintable():
        return
    n = uname(c)
    if not n:
        return
    return i,c,n,ca
    
it = range(30,200)
it = (allu(i) for i in it)
it = (t for t in it if t)
list(it)
```
